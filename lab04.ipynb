### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:

1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.



```sage
from sage.groups.perm_gps.permgroup_element import PermutationGroupElement


S = SymmetricGroup(4)
subgroups_generators = {
    "H1": [S("(1,2,3,4)")],
    "H2": [S("(1,3)"), S("(1,2,3,4)")],
    "H3": [S("(1,2)(3,4)"), S("(1,3)(2,4)")]
}

for name, gens in subgroups_generators.items():
    print(f"\nПідгрупа {name}")
    H = S.subgroup(gens)
    
    # 1
    h_order = H.order()
    h_index = S.order() / h_order
    print(f"1. Порядок підгрупи {name}: {h_order}")
    print(f"   Індекс підгрупи [S4 : {name}]: {h_index}")
    
    # 2
    left_cosets = [set(coset) for coset in S.cosets(H, side='left')]
    right_cosets = [set(coset) for coset in S.cosets(H, side='right')]
    
    print(f"2. Кількість лівих класів суміжності: {len(left_cosets)}")
    print(f"   Кількість правих класів суміжності: {len(right_cosets)}")
    
    # 3
    is_normal = H.is_normal(S)
    print(f"3. Чи є підгрупа {name} нормальною в S4? -> {is_normal}")
    
    # 4.
    if is_normal:
        print(f"4. Підгрупа {name} є нормальною")
        Q = S.quotient(H)
        print(f"   Структура факторгрупи: {Q.structure_description()}")
        print("   Таблиця Келі для факторгрупи:")
        print(Q.cayley_table())
        
    # 5
    else:
        print(f"5. Підгрупа {name} НЕ є нормальною. Шукаємо елемент 'a', де aH != Ha:")
        found_a = None
        for a in S:
            left_aH = set([a * h for h in H])
            right_Ha = set([h * a for h in H])
            
            if left_aH != right_Ha:
                found_a = a
                print(f"   Знайдено елемент a = {a}")
                print(f"   Лівий клас aH: {left_aH}")
                print(f"   Правий клас Ha: {right_Ha}")
                break

    
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
from sage.groups.perm_gps.permgroup_element import PermutationGroupElement

GL23 = GL(2, GF(3))
print(f"1. Порядок групи GL_2(Z_3): {GL23.order()}")

v1 = vector(GF(3), [1, 0])  # l1
v2 = vector(GF(3), [0, 1])  # l2
v3 = vector(GF(3), [1, 1])  # l3
v4 = vector(GF(3), [1, 2])  # l4
lines = [v1, v2, v3, v4]

def find_line_index(v):
    """Визначає, якій прямій пропорційний отриманий вектор v"""
    for idx, l in enumerate(lines):
        if matrix([v, l]).rank() == 1:
            return idx + 1
    return None

def get_permutation_for_matrix(A):
    """Знаходить підстановку з S_4 для матриці A"""
    images = []
    for v in lines:
        Av = A * v
        images.append(find_line_index(Av))
    S4 = SymmetricGroup(4)
    return S4(images).inverse()


S4 = SymmetricGroup(4)
images_of_generators = [get_permutation_for_matrix(gen) for gen in GL23.generators()]
Im_phi = S4.subgroup(images_of_generators)

print("2. Відображення phi успішно реалізовано через дію матриць на вектори прямих.")


# 3
is_homomorphism = True
for _ in range(10):
    A = GL23.random_element()
    B = GL23.random_element()
    print(f"{get_permutation_for_matrix(A * B)} ?= {get_permutation_for_matrix(A) * get_permutation_for_matrix(B)}")
    if get_permutation_for_matrix(A * B) != get_permutation_for_matrix(A) * get_permutation_for_matrix(B):
        is_homomorphism = False
        break

print(f"3. Чи є відображення phi гомоморфізмом? -> {is_homomorphism}")

# 4
N_elements = [A for A in GL23 if get_permutation_for_matrix(A) == S4.identity()]
N = GL23.subgroup(N_elements)

print(f"4. Порядок ядра N = ker(phi): {N.order()}")
print(f"   Елементи ядра N (скалярні матриці):")
for mat in N:
    print(f"   {mat.list()}")

print(f"   Порядок образу Im(phi) у групі S_4: {Im_phi.order()}")
print(f"   Чи збігається Im(phi) з усією групою S_4? -> {Im_phi == S4}")


# 5

# Оскільки функція GL23.quotient(N) не підтримується для матриць, 
# досліджуємо факторгрупу через її ізоморфний образ Im_phi в S_4.
print(f"5. За першою теоремою про гомоморфізми, факторгрупа GL_2(Z_3)/N ізоморфна образу Im(phi).")
print(f"   Порядок факторгрупи: {Im_phi.order()}")
print(f"   Структурний опис факторгрупи (образу): {Im_phi.structure_description()}")

```

### 5*. Проекції групи $SL_2(\mathbb{Z})$ на $SL_2(\mathbb{Z}_n)$ 

Розглянемо спеціальну лінійну групу $SL_2(\mathbb{Z})$, яка породжується двома матрицями:
$$S = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix},\qquad T = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}. $$

1. Задайте гомоморфізм $\rho_n: SL_2(\mathbb{Z}) \to SL_2(\mathbb{Z}_n)$, який редукує матриці за модулем $n$. 
2. Для значень $n\in\{2, 3, \ldots, 10\}$ перевірте, чи є $\rho_n$ сюр'єктивним, виконавши такі кроки:<br>
    a. Побудуйте групу $G_n = SL_2(\mathbb{Z}_n)$.<br>
    b. Побудуйте групу $H_n = \langle \rho_n(S), \rho_n(T) \rangle$.<br>
    c. Порівняйте порядки груп $H_n$ та $G_n$.



```sage


# Функція для перевірки сюр'єктивності проекції для заданого n
def check_sl2_surjectivity(n):
    Zn = IntegerModRing(n)
    
    Gn = SL(2, Zn)
    gn_order = Gn.order()
    
    # b
    S_mod = matrix(Zn, [[0, -1], [1, 0]])
    T_mod = matrix(Zn, [[1, 1], [0, 1]])
    
    # Будуємо підгрупу H_n, яку породжують ці дві редуковані матриці
    Hn = MatrixGroup([S_mod, T_mod])
    hn_order = Hn.order()
    
    # c
    is_surjective = (hn_order == gn_order)
    
    print(f"n = {n}:")
    print(f"Порядок повної групи SL_2(Z_{n}): {gn_order}")
    print(f"Порядок підгрупи H_n = <rho_n(S), rho_n(T)>: {hn_order}")
    print(f"Чи є гомоморфізм сюр'єктивним? -> {is_surjective}")
    print("-" * 50)
    
    return is_surjective


surjectivity_results = {}
for n in range(2, 11):
    surjectivity_results[n] = check_sl2_surjectivity(n)


all_surjective = all(surjectivity_results.values())
print(f"\nЧи є відображення редукції сюр'єктивним для всіх n від 2 до 10? -> {all_surjective}")
if all_surjective:
    print("Гіпотеза підтверджена, відображення редукції за модулем n є сюр'єктивним для будь-якого цілого n.")
```
