### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:

1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.



```sage
from collections import Counter, defaultdict


print("Group: Z*505")
print("Aplying Chinese remainder theorem to the group, we get that Z*505 is isomorphic to Z*5 x Z*101")
print("For every p - prime, multiplicative group Z*p will be isomorphic to additive group Zp-1")
print("in conclusion: Z*505 is isomorphic to Z4 x Z100")
print(70*"=","\n")

groups = [
    ("Z_400", [400]),
    ("Z_16 x Z_25", [16, 25]),
    ("Z_20 x Z_20", [20, 20]),
    ("Z_4 x Z_100", [4, 100]),
    ("Z_2 x Z_10 x Z_20", [2, 10, 20]),
    ("Z*505", [4, 100])
]

decomposition_dict = dict()

def find_elementary_divisors(cyclic_orders):
    return sorted(p^e for order in cyclic_orders for p, e in factor(order))
    

for name, invariants in groups:
    G = AdditiveAbelianGroup(invariants)
    elementary_divisors = find_elementary_divisors(invariants)
    p_group_decomposition = " x ".join(f"Z_{d}" for d in elementary_divisors)

    print(name)
    print("Group:", G)
    print("elementary divisors:", elementary_divisors)
    print("Decomposition into direct product:", p_group_decomposition,"\n")
    decomposition_dict[name] = p_group_decomposition
    
    order_counts = Counter(g.order() for g in G)
    
    for order, count in sorted(order_counts.items()):
        print(f"Order {order}: {count} elements")
    print(70*"-", "\n")
    

grouped = defaultdict(list)

for name, decomposition in decomposition_dict.items():
    grouped[decomposition].append(name)

duplicates = {val: keys for val, keys in grouped.items() if len(keys) > 1}
print(duplicates)   
```

2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```sage
#1
def count_abelian_groups(n):
    
    total_groups = 1
    for p, e in factor(n):
        total_groups *= number_of_partitions(e)
        
    return total_groups


print("Testing count_abelian_groups(n):")
for sample in [1, 4, 8, 12, 400]:
    print(f"n = {sample}: {count_abelian_groups(sample)} non-isomorphic abelian groups")
print(70 * "-", "\n")

#2
unique_group_ns = []
n = 1
while len(unique_group_ns) < 100:
    if count_abelian_groups(n) == 1:
        unique_group_ns.append(n)
    n += 1

print("2. First 100 natural numbers with a unique abelian group:")
print(unique_group_ns)
print("\nPrime factorizations of the first 100 found numbers:")
for val in unique_group_ns:
    print(f"n = {val:3d} -> factorization: {factor(val)}")

print(70 * "-", "\n")

# 3
print("Hypothesis:")
print("There exists a unique (up to isomorphism) abelian group of order n if and only if")
print("n is a square-free integer (i.e., in its prime factorization n = p1^e(1) * ... * pk^e(k),")
print("every exponent e(i) = 1.")
```

### 3. Дільники нуля та одиниці в скінченних кільцях
Розгляньте скінченні кільця $\mathbb{Z}_{30}$, $\mathbb{Z}_{12}\times\mathbb{Z}_{15}$, $M_3(\mathbb{Z}_3)$ та для кожного з них виконайте наступні кроки:
1. Знайдіть дільники нуля та їх кількість.
2. Знайдіть дільники одиниці та їх кількість.
3. Порівняйте отримані результати та зробіть висновок про взаємозв'язок між дільниками нуля та дільниками одиниці в скінченних кільцях.


```sage
R1 = Integers(30)
R2 = cartesian_product([Integers(12), Integers(15)])
R3 = MatrixSpace(Integers(3), 3, 3)

rings = [
    ("Z_30", R1),
    ("Z_12 x Z_15", R2),
    ("M_3(Z_3)", R3)
]

def check_element_type(x, R_set, zero_el, one_el):
    if x == zero_el:
        return "zero"
    

    if hasattr(x, "det"):  # Case for matrix ring M_3(Z_3)
        return "unit" if x.det() != 0 else "zero_divisor"
    elif isinstance(x, tuple):  # Case for Z_12 x Z_15
        a, b = x[0], x[1]
        if gcd(int(a), 12) == 1 and gcd(int(b), 15) == 1:
            return "unit"
        return "zero_divisor"
    else:  # Case for Z_30
        for y in R_set:
            if x * y == one_el and y * x == one_el:
                return "unit"
        return "zero_divisor"

for name, R in rings:
    print(f"Ring: {name}")
    elements = list(R)
    total_order = len(elements)
    print(f"Total elements in the ring: {total_order}")
    
    units = []
    zero_divisors = []
    
    for x in elements:
        el_type = check_element_type(x, elements, R.zero(), R.one())
        if el_type == "unit":
            units.append(x)
        elif el_type == "zero_divisor":
            zero_divisors.append(x)
            
    print(f"1. Number of non-zero zero divisors: {len(zero_divisors)}")
    print(f"   First 5 zero divisors: {zero_divisors[:5]} ...")
        
    print(f"2. Number of units (invertible elements): {len(units)}")
    print(f"   First 5 units: {units[:5]} ...")
        
    print(f"|Zero divisors (non-zero)| + |Units| + |{{0}}| = {len(zero_divisors)} + {len(units)} + 1 = {len(zero_divisors) + len(units) + 1}")
    print(70 * "-" + "\n")

# 3
print("3. Conclusion:")
print("In any finite ring R with unity (1 != 0), every non-zero element is either a unit(invertible element) or a zero divisor")
print("Therefore, the following partition holds: |R| = |U(R)| + |ZD(R) \\ {0}| + 1.")
print("- For Z_30: |R| = 30, |U(R)| = phi(30) = 8, non-zero zero divisors = 21.")
print("- For Z_12 x Z_15: |R| = 180, |U(R)| = phi(12)*phi(15) = 4*8 = 32, non-zero zero divisors = 147.")
print("- For M_3(Z_3): |R| = 3^9 = 19683, |U(R)| = |GL_3(F_3)| = (3^3-1)(3^3-3)(3^3-9) = 11232, non-zero zero divisors = 8450.")
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```sage
base_rings = [
    ("Q", QQ),
    ("Z_4", Integers(4)),
    ("Z_5", Integers(5))
]

for name, R in base_rings:
    print(f"Base ring R = {name}")
    Rx.<x> = PolynomialRing(R)
    f = x^4 + x + 2
    print(f"Polynomial f(x) = {f}")
    
    # 1
    I = Rx.ideal(f)
    print(f"1. Ideal I = {I}")
    
    # 2
    Q_ring.<x_bar> = Rx.quotient(I)
    order_Q = Q_ring.cardinality()
    print(f"2. Quotient ring R[x]/I: {Q_ring}")
    print(f"Order of R[x]/I: {order_Q}")
    
    # 3
    is_field = Q_ring.is_field() if R.is_field() else False
    print(f"3. Is R[x]/I a field? {is_field}")
    
    # 4
    print("4. Zero divisors in R[x]/I:")
    if R == QQ:
        print("Since R[x]/I is a field, there are NO non-zero zero divisors.")
    else:
        elements = list(Q_ring)
        zero_el = Q_ring(0)
        one_el = Q_ring(1)
        zero_divisors = []
        
        for el in elements:
            if el == zero_el:
                continue
            is_unit = False
            for y in elements:
                if el * y == one_el:
                    is_unit = True
                    break
            if not is_unit:
                zero_divisors.append(el)
                
        print(f"Number of non-zero zero divisors: {len(zero_divisors)}")
        print(f"First 5 non-zero zero divisors: {zero_divisors[:5]}")
        

    print("5. Factorization of f(x) over R:")
    if R.is_field():
        fact = f.factor()
        print(f"f(x) = {fact}")
        roots = f.roots(multiplicities=False)
        print(f"Roots of f(x) in {name}: {roots}")
        if roots:
            print("Linear factors (x - r) correspond directly to zero divisors in R[x]/I,")
            print("because f(x) = (x - r)*g(x) implies (x_bar - r)*g(x_bar) = 0 in R[x]/I.")
        else:
            print("f(x) has no roots and no proper factors over Q, so R[x]/I has no non-zero zero divisors.")
    else:
        roots_Z4 = [r for r in R if f(r) == 0]
        print(f"{name} is not an Unique Factorization Domain​(факторіальне кільце), so unique factorization does not apply.")
        print(f"Roots of f(x) in {name}: {roots_Z4}")
        if not roots_Z4:
            print("f(x) has NO linear factors (x - r) over Z_4.")
            print("Even without linear factors of f(x), R[x]/I still has many zero divisors")
            print("because the base ring Z_4 already has zero divisors for example 2 * 2 = 0 in Z_4[x]/I")

    print("6. Invertibility of x + I in R[x]/I:")
    try:
        inv_x = x_bar^(-1)
        print(f"   x + I IS invertible Its inverse is: {inv_x}")
        print(f"   Check: (x + I) * ({inv_x}) = {x_bar * inv_x}")
    except Exception as e:
        print(f"   x + I is NOT invertible in R[x]/I in {name}")
        
    print(70 * "-" + "\n")
```
